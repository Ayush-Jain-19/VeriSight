# 01. Exploratory Data Analysis & Manifest Inspection

This notebook inspects the FakeAVCeleb dataset structure, demographic distribution, category balance, and identity groupings.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

manifest_path = Path('../data/manifest.csv')
if manifest_path.exists():
    df = pd.read_csv(manifest_path)
    print(f'Total samples: {len(df)}')
    print(f'Unique identities: {df["identity_id"].nunique()}')
    display(df.head())
else:
    print('manifest.csv not found. Run scripts/build_manifest.py first.')
```

```python
# Category distribution
if 'df' in locals():
    print(df['category'].value_counts())
    df['category'].value_counts().plot(kind='bar', title='Category Distribution')
    plt.show()
```
